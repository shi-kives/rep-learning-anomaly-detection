#### Missing Files

```python
def calculate_missing(df):
    return sum([0 if Path(file_path).is_file() else 1 for file_path in df['Path_to_PNG']])

print("missing files in train: ", calculate_missing(train))
print("missing files in val: ", calculate_missing(val))
print("missing files in test: ", calculate_missing(test))
```

#### Duplicate Files

```python
def find_duplicates(df: pd.DataFrame, dir_path: Path):
    files_in_dir = set(f.name for f in dir_path.rglob("*.png"))
    files_in_df = set(Path(p).name for p in df['Path_to_PNG'] )

    return len(files_in_df - files_in_dir)

print("duplicate files in train: ", find_duplicates(train, Path('datasets/MVTecAD2/can/train/good')))
print("duplicate files in val: ", find_duplicates(val, Path('datasets/MVTecAD2/can/validation/good')))
```

```python
print("duplicates in train df: ", train['Path_to_PNG'].duplicated().sum())
print("duplicates in val df: ", val['Path_to_PNG'].duplicated().sum())
print("duplicates in test df: ", test['Path_to_PNG'].duplicated().sum())
```

#### Normal-Anomalous Distribution

```python
def plot_anomaly_distribution(df: pd.DataFrame, mode):
    distribution = df['Label'].map({0: 'Normal', 1: 'Anomalous'}).value_counts()

    plt.figure(figsize=(4,4))
    plt.pie(distribution, labels = distribution.index, autopct="%1.1f%%")
    plt.title(f"Anomaly Distribution Chart for {mode.capitalize()}")
    plt.show()

plot_anomaly_distribution(train, "train")
plot_anomaly_distribution(val, "val")
plot_anomaly_distribution(test, "test")
```

#### Domain Distribution

```python
def plot_domain_distribution(df: pd.DataFrame, mode: str):
    distribution = df['Domain'].value_counts().astype(int).to_dict()

    plt.figure(figsize=(4,4))
    plt.pie(distribution.values(), labels = distribution.keys(), autopct="%1.1f%%")
    plt.title(f"Domain Distribution for {mode.capitalize()}")
    plt.show()

plot_domain_distribution(train, "train")
plot_domain_distribution(val, "val")
plot_domain_distribution(test, "test")
```

### Confirming No Overlap

```python
train_paths = set(train['Path_to_PNG'])
val_paths = set(val['Path_to_PNG'])
test_paths = set(test['Path_to_PNG'])

all3 = train_paths & val_paths & test_paths
print("number of common images in train, val, test: ", len(all3))

train_val = train_paths & val_paths
print("number of common images in train, val: ", len(train_val))

train_test = train_paths & test_paths
print("number of common images in train, test: ", len(train_test))

test_val = test_paths & val_paths
print("number of common images in test, val: ", len(test_val))
```
