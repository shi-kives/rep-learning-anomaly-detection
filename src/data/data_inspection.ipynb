## Inspecting DataLoader & ImageDataset

```python
from src.data.data_loader import ImageDataset
from torch.utils.data import DataLoader
```

```python
def inspect_dataset(csv_path: str, mode: str):

    dataset = ImageDataset(csv_path)
    print(f"sample from {mode} dataset of index 0:\n", dataset[0])

    print(f"\nnumber of samples in {mode} dataset: ", len(dataset))
    print("shape of image torch tensor: ", dataset[0][1].shape)

    shuffle = True if mode == 'train' else False
    loader = DataLoader(dataset, batch_size = 32, shuffle = shuffle)
    print(f"\nnumber of batches in {mode} loader: ", len(loader))

    for batch_sno, batch_images, batch_labels in loader:
        print("\nsample batch numbers: ", batch_sno[:5])
        print("sample batch image tensors:\n", batch_images[0])
        print("sample batch labels: ", batch_labels[:5])
        break
```

```python
inspect_dataset('results/artifacts/train_manifest.csv', 'train')
```

```python
inspect_dataset("results/artifacts/validation_manifest.csv", "validation")
```

```python
inspect_dataset("results/artifacts/test_public_manifest.csv", "test")
```
